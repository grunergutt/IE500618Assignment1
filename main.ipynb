```python
import os
import pandas as pd
import matplotlib as plt

BASE_DIR = os.path.join(os.getcwd())
PATH_DATA = os.path.join(BASE_DIR, "data")
PATH_AMESHOUSING = os.path.join(PATH_DATA, "AmesHousing.csv")

df = pd.read_csv(PATH_AMESHOUSING)
print(df.info())
```

```python
print(df.isna().sum())
```

```python
import matplotlib.pyplot as plt

# Count missing values per column and filter only columns with missing values
missing_values = df.isna().sum()
missing_values_filtered = missing_values[missing_values > 0].sort_values(ascending=False)

# Create the plot
plt.figure(figsize=(14, 6))
missing_values_filtered.plot(kind='bar', color='steelblue')
plt.title('Missing Values per Column', fontsize=14, fontweight='bold')
plt.xlabel('Columns', fontsize=12)
plt.ylabel('Count of Missing Values (NaN)', fontsize=12)
plt.xticks(rotation=90)
plt.tight_layout()
plt.grid(axis='y', alpha=0.3)
plt.show()

print(f"\nTotal missing values per column:\n{missing_values_filtered}")
```
