```python
import os
import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = os.path.join(os.getcwd())
PATH_DATA = os.path.join(BASE_DIR, "data")
PATH_AMESHOUSING = os.path.join(PATH_DATA, "AmesHousing.csv")

df = pd.read_csv(PATH_AMESHOUSING)
print("\nInfo:")
print(df.info())
```

```python
df_sale_price = df[df.columns[81]]
print("\nSale price:")
print(df_sale_price.sort_values(ascending=False))

plt.grid(True)
plt.hist(df_sale_price, bins=100)
plt.show()
```

```python
df_year_built = df[df.columns[20]]
print("\nYear built:")
print(df_year_built.sort_values(ascending=False))

plt.grid(True)
plt.hist(df_year_built, bins=100)
plt.show()
```
