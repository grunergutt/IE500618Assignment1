```python
# Import block
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import altair as alt
from config import PATH_AMESHOUSING
```

```python
# Functions
def standard_figure_size():
    plt.figure(figsize=(12, 6))
    plt.grid(True)
```

```python
df = pd.read_csv(PATH_AMESHOUSING)

print("Dataframe:")
print(df)
```

```python
print("\nDataframe Info:")
print(df.info())
```

```python
df_lot_area = df[df.columns[5]]
df_year_built = df[df.columns[20]]
df_sale_price = df[df.columns[81]]

fig, axs = plt.subplots(3, 1)

axs[0].hist(df_lot_area, bins=100)
axs[0].set_title("Lot Area")
axs[1].hist(df_year_built, bins=100)
axs[1].set_title("Year Built")
axs[1].set_xlabel("Year")
axs[2].hist(df_sale_price, bins=100)
axs[2].set_title("Sale Price:")
axs[2].set_xlabel("Price USD")

for ax in axs:
    ax.grid(True)

plt.tight_layout()
# plt.hist(df_year_built, bins=100);
```

```python
correlation = (df.corr(numeric_only=True)["Year Built"]).sort_values(ascending=False).round(2)

correlation
```

```python
# Heatmap for correlation
plt.figure(figsize=(40, 24))
sns.heatmap(data=df.corr(numeric_only=True), cmap="Spectral", annot=True, cbar=True)
plt.title("Correlation of the whole dataframe", fontsize=24)
```

```python
sns.set_theme(style="whitegrid")

g = sns.JointGrid(
    data=df,
    x="Gr Liv Area",
    y="SalePrice",
    space=0
)

g.plot_joint(sns.kdeplot, fill=True)
g.plot_marginals(sns.histplot, bins=25)
g.set_axis_labels("Ground Living Area (ft$^2$)", "Sale Price (USD)")
g.ax_joint.grid(True)
```
