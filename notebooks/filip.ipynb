```python
SaPr = df["SalePrice"]

plt.figure(figsize=(10, 6))
plt.hist(SaPr, bins=30, edgecolor="black")
plt.xlabel("Sale Price")
plt.ylabel("Frequency")
plt.title("Histogram of Sale Price")
plt.show()

plt.figure(figsize=(10, 6))
plt.grid(True)
box = plt.boxplot(SaPr, vert=False, flierprops={'markersize': 2, 'markeredgecolor': 'blue'},
                   patch_artist=True)
for patch in box['boxes']:
    patch.set_facecolor('lightblue')
plt.xlabel("Sale Price")
plt.ylabel("Visualization of usual and unusual spenders")
plt.title("Boxplot of Sale Price")
plt.show()
```
